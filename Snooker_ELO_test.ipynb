```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

```python
filePath = r"D:\Snooker_Selenium\H2H32\players_score_history.csv" # modify this path
df = pd.read_csv(filePath)
df['date'] = pd.to_datetime(df['date'])
players_to_plot = ['Judd Trump', 'Zhao Xintong', 'Mark Selby']

plt.figure(figsize=(12, 6))

# 4. Plot each player's trajectory
for player in players_to_plot:
    player_data = df[df['name'] == player].sort_values(by='date')

    # We use a step plot so the score stays flat between matches,
    # and jumps up/down exactly on the match date.
    plt.step(player_data['date'], player_data['score'], where='post', label=player, linewidth=2)

# 5. Formatting the chart
plt.title("Snooker Form Evolution (2023 - Present)", fontsize=16)
plt.xlabel("Date", fontsize=12)
plt.ylabel("Form Points (Elo)", fontsize=12)
plt.legend(loc="upper left")
plt.grid(True, linestyle='--', alpha=0.7)

# Format the x-axis to look nice with dates
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
plt.gca().xaxis.set_major_locator(mdates.MonthLocator(interval=2))
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
